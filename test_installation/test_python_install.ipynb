# Python Installation Test

This notebook checks whether your Python installation and Jupyter setup are working correctly.
It works on **Windows**, **macOS**, and **Linux**, and in both plain Python environments and **conda** environments.

**How to use it:**
1. Run each cell in order (click a cell, then press `Shift + Enter`).
2. Look for `[OK]` messages. If you see any `[FAIL]` messages, show them to your instructor.


````python
import sys
import os
import platform
import subprocess

def check(label, condition, extra=""):
    status = "[OK]" if condition else "[FAIL]"
    print(f"{status} {label}{(' - ' + extra) if extra else ''}")
    return condition

all_good = True
````

## 1. Python version and operating system

````python
version = sys.version.split()[0]
major, minor = sys.version_info[0], sys.version_info[1]
ok = (major == 3 and minor >= 8)
all_good &= check("Python version >= 3.8", ok, f"found {version}")

os_name = platform.system()  # 'Windows', 'Darwin' (Mac), or 'Linux'
check("Operating system detected", True, f"{os_name} ({platform.platform()})")
````

## 2. Basic computation and standard library

````python
result = sum(range(1, 101))
all_good &= check("Basic computation works", result == 5050, f"sum(1..100) = {result}")

try:
    import json
    import datetime
    import math
    all_good &= check("Standard library imports work", True)
except ImportError as e:
    all_good &= check("Standard library imports work", False, str(e))
````

## 3. File read/write (cross-platform)

````python
try:
    test_path = os.path.join(os.getcwd(), "_install_test_tmp.txt")
    with open(test_path, "w") as f:
        f.write("hello from the notebook")
    with open(test_path, "r") as f:
        content = f.read()
    os.remove(test_path)
    all_good &= check("File read/write works", content.startswith("hello"))
except OSError as e:
    all_good &= check("File read/write works", False, str(e))
````

## 4. Check that `pip` works

This checks `pip` using the *same* Python that is running this notebook, which is the safest
way to confirm pip will install packages into the right environment (plain Python, venv, or conda).

````python
try:
    pip_check = subprocess.run(
        [sys.executable, "-m", "pip", "--version"],
        capture_output=True, text=True, timeout=15
    )
    pip_ok = pip_check.returncode == 0
    pip_info = pip_check.stdout.strip() if pip_ok else pip_check.stderr.strip()
    all_good &= check("pip is available", pip_ok, pip_info)
except Exception as e:
    all_good &= check("pip is available", False, str(e))

check("Python executable in use", True, sys.executable)
````

## 5. Try an actual `pip install`

The cell below installs a small, harmless package (`wheel`) to confirm that `pip install` works
end-to-end on your system or in your active conda environment.

- In Jupyter, `%pip install ...` (the "magic" command) automatically installs into the Python
  environment powering this notebook. This is the safest way to do it, and works the same on
  Windows and Mac.
- If you'd rather test it from a terminal instead, close/ignore this cell and run this in your
  terminal or Anaconda Prompt:
  ```
  pip install wheel
  ```

````python
%pip install --quiet wheel
print("If you see no red error text above, pip install works correctly.")
````

## 6. Final summary

````python
print("=" * 60)
if all_good:
    print("SUCCESS: Your Python installation looks good!")
else:
    print("Some checks FAILED. Please show this output to your instructor.")
print("=" * 60)
````
